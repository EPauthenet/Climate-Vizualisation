```python
#Check for missing dates
import os
from datetime import datetime, timedelta
import glob
```

```python
# Function to convert string to datetime
def str_to_date(date_str):
    return datetime.strptime(date_str, '%Y-%m-%d')

# Function to convert datetime to string
def date_to_str(date):
    return date.strftime('%Y-%m-%d')
```

```python
# Define list expected dates
data_path = '/Users/epauthen/Documents/Database/Antarctic_Sea_Ice/v3p0/'
file_list = glob.glob(data_path + '*/*/ice_conc_sh_*.nc')

expected_dates = []
for file_path in file_list:
    date_str = file_path.split('/')[-1].split('_')[-1][:8]
    formatted_date = f"{date_str[:4]}-{date_str[4:6]}-{date_str[6:]}"
    expected_dates.append(str_to_date(formatted_date))

# Get the list of existing files and extract dates
existing_files = os.listdir('/Users/epauthen/Documents/Database/Antarctic_Sea_Ice/v3p0_shape15/')
existing_dates = [str_to_date(file.split('_')[1].split('.')[0]) for file in existing_files if file.startswith('Shape15')]
print(len(existing_dates))
print(14834-len(existing_dates))

# Find missing dates
missing_dates = [date_to_str(date) for date in expected_dates if date not in existing_dates]
len(missing_dates)

# Write missing dates to a file
output_file = 'missing_dates.txt'
with open(output_file, 'w') as f:
    for date in missing_dates:
        f.write(date + '\n')
```

```python
# Open and read the missing_dates.txt file
with open('missing_dates.txt', 'r') as f:
    missing_dates = f.read().splitlines()

# Define data path and list of files
data_path = '/Users/epauthen/Documents/Database/Antarctic_Sea_Ice/v3p0/'
file_list = glob.glob(data_path + '*/*/ice_conc_sh_*.nc')

# Find file paths corresponding to missing dates
missing_files = []
for file_path in file_list:
    date_str = file_path.split('/')[-1].split('_')[-1][:8]
    formatted_date = f"{date_str[:4]}-{date_str[4:6]}-{date_str[6:]}"
    if formatted_date in missing_dates:
        missing_files.append(file_path)

# Write missing dates to a file
output_file = 'missing_files.txt'
with open(output_file, 'w') as f:
    for file in missing_files:
        f.write(file + '\n')
```
